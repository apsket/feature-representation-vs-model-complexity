```python
from PIL import Image

# Open your images
img1 = Image.open('figures/sample_synthetic_elliptical_dataset.png')
img2 = Image.open('synthetic_elliptical_summary.png')

# Define the new dimensions (combine widths, use the max height)
total_width = img1.width + img2.width
max_height = max(img1.height, img2.height)

# Create a new blank image with a transparent or white background
new_img = Image.new('RGBA', (total_width, max_height), (255, 255, 255, 0))

# Paste the images side-by-side
new_img.paste(img1, (0, 0))
new_img.paste(img2, (img1.width, 0))

# Save the result
new_img.save('combined_graphs.png')
```

```python
from PIL import Image

# Open images and store them in a list
image_files = [
    'figures/synthetic_circular_small_dataset.png',
    'figures/synthetic_circular_medium_dataset.png'
    'figures/synthetic_circular_large_dataset.png',
]
images = [Image.open(img) for img in image_files]

# Calculate the total width (sum of all) and max height
total_width = sum(img.width for img in images)
max_height = max(img.height for img in images)

# Create a new blank image with a transparent background
new_img = Image.new('RGBA', (total_width, max_height), (255, 255, 255, 0))

# Paste the images side-by-side using a tracking variable for the X coordinate
current_x = 0
for img in images:
    new_img.paste(img, (current_x, 0))
    current_x += img.width  # Shift the starting point for the next image

# 5. Save the result
new_img.save('combined_3_graphs.png')
```

```python
from PIL import Image, ImageOps

# 1. Open your 4 images
img1 = Image.open('figures/logi_regr_synthetic_circular_polar_dataset.png')
img2 = Image.open('figures/logi_regr_synthetic_circular_r_dataset.png')
img3 = Image.open('figures/logi_regr_synthetic_circular_curvilinear_dataset.png')
img4 = Image.open('figures/svm_synthetic_circular_cartesian_dataset.png')

# 2. Use the first image's size as the standard/target size
target_width = img1.width
target_height = img1.height
target_size = (target_width, target_height)

# 3. Scale images keeping aspect ratio, padding with white where needed
# (Using Resampling.LANCZOS for high quality and centering by default)
pad_color = (255, 255, 255, 255) # White background (RGBA)

img1_res = ImageOps.pad(img1, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img2_res = ImageOps.pad(img2, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img3_res = ImageOps.pad(img3, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img4_res = ImageOps.pad(img4, target_size, method=Image.Resampling.LANCZOS, color=pad_color)

# 4. Calculate total canvas size (exactly 2x the target dimensions)
total_width = target_width * 2
total_height = target_height * 2

# 5. Create the canvas (using white background instead of transparent to match padding)
grid_img = Image.new('RGBA', (total_width, total_height), (255, 255, 255, 255))

# 6. Paste the padded images into the 2x2 grid
grid_img.paste(img1_res, (0, 0))                           # Top-Left
grid_img.paste(img2_res, (target_width, 0))                # Top-Right
grid_img.paste(img3_res, (0, target_height))               # Bottom-Left
grid_img.paste(img4_res, (target_width, target_height))     # Bottom-Right

# 7. Save the final uniform grid
grid_img.save('uniform_2x2_grid.png')
```
