```python
from PIL import Image

# Open images and store them in a list
image_files = [
    'figures/synthetic_circular_small_dataset.png',
    'figures/synthetic_circular_medium_dataset.png',
    'figures/synthetic_circular_large_dataset.png'
]
images = [Image.open(img) for img in image_files]

# Calculate the total width (sum of all) and max height
total_width = sum(img.width for img in images)
max_height = max(img.height for img in images)

# Create a new blank image with a transparent background
new_img = Image.new('RGBA', (total_width, max_height), (255, 255, 255, 0))

# Paste the images side-by-side using a tracking variable for the X coordinate
current_x = 0
for img in images:
    new_img.paste(img, (current_x, 0))
    current_x += img.width  # Shift the starting point for the next image

# Save the result
new_img.save('combined_3_graphs.png')
```

```python
from PIL import Image, ImageOps

# Open 4 images
img1 = Image.open('figures/logi_regr_polar_synthetic_elliptical_dataset.png')
img2 = Image.open('figures/logi_regr_curvilinear_synthetic_elliptical_dataset.png')
img3 = Image.open('figures/svm_synthetic_elliptical_dataset.png')
img4 = Image.open('figures/logi_regr_quad_synthetic_elliptical_dataset.png')

# Use the first image's size as the standard/target size
target_width = img1.width
target_height = img1.height
target_size = (target_width, target_height)

# Scale images keeping aspect ratio, padding with white where needed
# (Using Resampling.LANCZOS for high quality and centering by default)
pad_color = (255, 255, 255, 255) # White background (RGBA)

img1_res = ImageOps.pad(img1, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img2_res = ImageOps.pad(img2, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img3_res = ImageOps.pad(img3, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img4_res = ImageOps.pad(img4, target_size, method=Image.Resampling.LANCZOS, color=pad_color)

# Calculate total canvas size (exactly 2x the target dimensions)
total_width = target_width * 2
total_height = target_height * 2

# Create the canvas (using white background instead of transparent to match padding)
grid_img = Image.new('RGBA', (total_width, total_height), (255, 255, 255, 255))

# Paste the padded images into the 2x2 grid
grid_img.paste(img1_res, (0, 0))                           # Top-Left
grid_img.paste(img2_res, (target_width, 0))                # Top-Right
grid_img.paste(img3_res, (0, target_height))               # Bottom-Left
grid_img.paste(img4_res, (target_width, target_height))     # Bottom-Right

# Save the final uniform grid
grid_img.save('non_radial_symmetric_pre_summary.png')
```

```python
from PIL import Image, ImageOps

# Open 6 images
img1 = Image.open('figures/synthetic_non_rad_dataset.png')
img2 = Image.open('figures/logi_regr_polar_synthetic_non_rad_dataset.png')
img3 = Image.open('figures/logi_regr_curvilinear_synthetic_non_rad_dataset.png')
img4 = Image.open('figures/svm_synthetic_non_rad_dataset.png')
img5 = Image.open('figures/svm_synthetic_non_rad_polar_dataset.png')
img6 = Image.open('figures/logi_regr_synthetic_non_rad_quad_dataset.png')

# Use the first image's size as the standard/target size
target_width = img2.width
target_height = img2.height
target_size = (target_width, target_height)

# Scale images keeping aspect ratio, padding with white where needed
# (Using Resampling.LANCZOS for high quality and centering by default)
pad_color = (255, 255, 255, 255) # White background (RGBA)

img1_res = ImageOps.pad(img1, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img2_res = ImageOps.pad(img2, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img3_res = ImageOps.pad(img3, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img4_res = ImageOps.pad(img4, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img5_res = ImageOps.pad(img5, target_size, method=Image.Resampling.LANCZOS, color=pad_color)
img6_res = ImageOps.pad(img6, target_size, method=Image.Resampling.LANCZOS, color=pad_color)

# Calculate total canvas size (exactly 2x the target dimensions)
total_width = target_width * 3
total_height = target_height * 2

# Create the canvas (using white background instead of transparent to match padding)
grid_img = Image.new('RGBA', (total_width, total_height), (255, 255, 255, 255))

# Paste the padded images into the 2x2 grid
grid_img.paste(img1_res, (0, 0))                             # Top-Left
grid_img.paste(img2_res, (target_width, 0))                  # Top-Center
grid_img.paste(img3_res, (2*target_width, 0))                # Top-Right
grid_img.paste(img4_res, (0, target_height))                 # Bottom-Left
grid_img.paste(img5_res, (target_height, target_height))     # Bottom-Center
grid_img.paste(img6_res, (2*target_width, target_height))    # Bottom-Right

# Save the final uniform grid
grid_img.save('non_radial_symmetric_summary.png')
```
